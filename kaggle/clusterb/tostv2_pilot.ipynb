```python
import os, glob, shutil
hits = glob.glob('/kaggle/input/**/tests/common_smoke.py',
                 recursive=True)
assert hits, 'common_smoke no encontrado; input=' + repr(os.listdir('/kaggle/input'))
src = os.path.dirname(os.path.dirname(hits[0]))
print('ARTEFACTOS EN:', src)
dst = '/kaggle/working/artifacts'
if not os.path.exists(dst):
    shutil.copytree(src, dst)
os.chdir(dst)
open('/kaggle/working/artifacts/tests/__init__.py', 'a').close()
os.makedirs('outputs/wave_mem', exist_ok=True)
```
